```python
import sys
sys.path.insert(0, '/content/') # Ensure /content/ is in sys.path

%matplotlib inline
from grid_world import GridWorldEnv # Corrected import statement
import numpy as np

env = GridWorldEnv(grid_size=8, start_pos=(0,0), goal_pos=(7,7))
obs, _ = env.reset()
print("Environment ready!")
print("State shape:", obs.shape)
print("Actions available:", env.n_actions)
```

```python
from q_learning import QLearningAgent

agent = QLearningAgent(
    n_states=env.n_states,
    n_actions=env.n_actions
)
history = agent.train(env, n_episodes=1000)
```

```python
from random_agent import RandomAgent

env2 = GridWorldEnv(grid_size=8, start_pos=(0,0), goal_pos=(7,7))
random_agent = RandomAgent(n_actions=env2.n_actions)
random_history = random_agent.run(env2, n_episodes=1000)
```

```python
from random_agent import RandomAgent

env2 = GridWorldEnv(grid_size=8, start_pos=(0,0), goal_pos=(7,7))
random_agent = RandomAgent(n_actions=env2.n_actions)
random_history = random_agent.run(env2, n_episodes=1000)
```

```python
import matplotlib.pyplot as plt

# Extract rewards from history
q_learning_rewards = history['rewards']
random_agent_rewards = random_history['rewards']

# Create episode numbers for the x-axis
episodes = range(len(q_learning_rewards))

plt.figure(figsize=(12, 6))
plt.plot(episodes, q_learning_rewards, label='Q-Learning Agent')
plt.plot(episodes, random_agent_rewards, label='Random Agent')

plt.xlabel('Episode')
plt.ylabel('Total Reward')
plt.title('Comparison of Total Rewards per Episode')
plt.legend()
plt.grid(True)
plt.show()
```

```python
import os
import json
import numpy as np

os.makedirs("models", exist_ok=True)

# Manual save — bypasses any type issues
data = {
    "q_table": agent.q_table.tolist(),
    "n_states": int(agent.n_states),
    "n_actions": int(agent.n_actions),
    "learning_rate": float(agent.lr),
    "gamma": float(agent.gamma),
    "epsilon": float(agent.epsilon),
    "episode_rewards": [float(r) for r in agent.episode_rewards[-100:]]
}

with open("models/q_table.json", "w") as f:
    json.dump(data, f, indent=2)

print("Model saved successfully!")
```

```python
eval_env = GridWorldEnv(grid_size=8, start_pos=(0,0), goal_pos=(7,7))
obs, _ = eval_env.reset()
state = eval_env.get_state_index()

while True:
    action = agent.select_action(state, training=False)
    _, _, term, trunc, info = eval_env.step(action)
    state = eval_env.get_state_index()
    if term or trunc:
        break

eval_env.visualize_path(
    q_table=agent.q_table,
    title="Q-Learning: Trained Robot Path"
)
```

```python
import importlib.util

spec = importlib.util.spec_from_file_location(
    "dqn_agent",
    "/content/robot_nav_rl/agents/dqn_agent.py"
)
dqn_module = importlib.util.module_from_spec(spec)
spec.loader.exec_module(dqn_module)

print("✅ DQN agent loaded!")
```

```python
env_dqn = GridWorldEnv(grid_size=8, start_pos=(0,0), goal_pos=(7,7))

dqn_agent = dqn_module.DQNAgent(
    state_size    = env_dqn.observation_space.shape[0],
    n_actions     = int(env_dqn.n_actions), # Explicitly cast to int
    learning_rate = 0.001,
    gamma         = 0.99,
    epsilon       = 1.0,
    epsilon_min   = 0.01,
    epsilon_decay = 0.995,
    batch_size    = 64
)

dqn_history = dqn_agent.train(env_dqn, n_episodes=500, verbose=True, verbose_every=50)
```

```python
plotting.plot_comparison({
    "q_learning": history,
    "random":     random_history,
    "dqn":        dqn_history
})
```

```python
plotting.print_comparison_table({
    "q_learning": result_ql,
    "random": {
        "mean_reward":  sum(random_history["rewards"])/len(random_history["rewards"]),
        "std_reward":   float(np.std(random_history["rewards"])),
        "mean_length":  sum(random_history["lengths"])/len(random_history["lengths"]),
        "success_rate": sum(1 for r in random_history["rewards"] if r > 0) / len(random_history["rewards"])
    },
    "dqn": {
        "mean_reward":  float(np.mean(dqn_history["rewards"][-100:])),
        "std_reward":   float(np.std(dqn_history["rewards"][-100:])),
        "mean_length":  float(np.mean(dqn_history["lengths"][-100:])),
        "success_rate": dqn_history["final_success_rate"]
    }
})
```
